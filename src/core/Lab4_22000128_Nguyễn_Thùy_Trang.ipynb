```python
import gensim.downloader as api
import numpy as np

# Tải mô hình pre-trained
def load_model(model_path = model_path):
  print("Đang tải mô hình glove-wiki-gigaword-50...")
  try:
      model = api.load(model_path)
      print("Tải mô hình thành công! ")
  except Exception as e:
      print(f"Lỗi khi tải mô hình: {e}")
      model = None
load_model()
```

```python
def get_embedding(word: str, model):
  if model is not None:
    try:
        vector_of_word = model[word]
        return vector_of_word
    except KeyError:
        print(f"Từ '{word}' không có trong vốn từ vựng của mô hình.")

word = ""
vector_of_word = get_embedding(word, model)
print(f"Vector của từ '{word}': {vector_of_word}")
```

```python
def get_similarity(word1, word2, model):
  try:
    similarity= model.similarity(word1, word2)
    return similarity
  except Exception as e:
      print(f"Lỗi khi tính độ tương đồng: {e}")

word2 = ""
similarity = get_similarity(word, word2, model)
print(f"Độ tương đồng giữa '{word}' và '{word2}': {similarity:.4f}")
```

```python
def most_similar(k, word, model):
  try:
    most_similar_words = model.most_similar(word, topn=k)
    return most_similar_words
  except Exception as e:
      print(f"Lỗi khi tìm các từ đồng nghĩa: {e}")

word = ""
k = 10
print(f"\nCác từ đồng nghĩa/liên quan nhất với '{word}' (Top {k}):")
for word, similarity in most_similar(word, model):
        print(f" - {word}: {similarity:.4f}")
```
